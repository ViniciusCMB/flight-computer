```python
import matplotlib.pyplot as plt
import pandas as pd
import csv
```

```python

data_cleaned = []

with open('voo_thonyan.txt', 'r', encoding='utf-8') as f:
    for line in f:
        data = line.strip()
        
        # Ignora linhas que começam com '['
        if data.startswith('['):
            continue 
            
        if data:
            # .split(',') transforma a string "#51,2018596,..." 
            # em uma lista de elementos: ["#51", "2018596", "..."]
            linhas_separadas = data.split(',')
            data_cleaned.append(linhas_separadas)

# Salvando no CSV com as colunas separadas corretamente
with open("cleaned.csv", "w", newline="", encoding="utf-8") as arquivo:
    escritor = csv.writer(arquivo)
    escritor.writerows(data_cleaned)
```

```python
df = pd.read_csv("cleaned.csv", sep=",", header=None)
```

```python
columns = ['ID', 'millis', 'pckt', 'alt', 'temp', 'humi', 'press', 'gx', 'gy', 'gz', 'ax', 'ay', 'az', 'vz', 'maxalt', 'state', 'hora', 'data', 'gps_alt', 'lat', 'lon', 'sat', 'pqd', 'rssi']

df.columns = columns

df.head()
```

```python
plt.plot(df['millis'], df['alt'])
plt.xlabel('Millis')
plt.ylabel('Altitude')
plt.title('Altitude vs Millis')
plt.show()
```

```python
plt.plot(df['millis'], df['ax'])
plt.plot(df['millis'], df['ay'])
plt.plot(df['millis'], df['az'])
plt.xlabel('Millis')
plt.ylabel('Acceleration')
plt.title('Acceleration vs Millis')
plt.legend(['ax', 'ay', 'az'])
plt.show()
```
